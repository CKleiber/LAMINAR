```python
import matplotlib.pyplot as plt
```

```python
def T1(moves = ['u', 'r', 'd', 'l']):
    return moves[1] + moves[0] + moves[3] + moves[2]

def T2(moves = ['u', 'r', 'd', 'l']):
    return moves[3] + moves[2] + moves[1] + moves[0]

def hilbert_curve(n, moves = ['u', 'r', 'd', 'l']):
    # basic unit is:
    # u, r, d 
    # -> moves[0], moves[1], moves[2]

    if n == 0:
        return []
    
    else:
        arr = []

        moves = T1(moves)
        arr += (hilbert_curve(n - 1, moves))
        moves = T1(moves)

        arr += (moves[0])

        arr += (hilbert_curve(n - 1, moves))

        arr += (moves[1])

        arr += (hilbert_curve(n - 1, moves))

        arr += (moves[2])

        moves = T2(moves)
        arr += (hilbert_curve(n - 1, moves))
        moves = T2(moves)

        return arr
```

```python
hilbert_curve(n=3)
```

```python
def move(position, direction):
    """
    Move the position in the given direction.
    """
    if direction == 'u':
        return (position[0], position[1] + 1)
    elif direction == 'd':
        return (position[0], position[1] - 1)
    elif direction == 'l':
        return (position[0] - 1, position[1])
    elif direction == 'r':
        return (position[0] + 1, position[1])
    else:
        raise ValueError("Invalid direction: {}".format(direction))
```

```python
positions = [(0, 0)]
directions = hilbert_curve(6)
for direction in directions:
    new_position = move(positions[-1], direction)
    positions.append(new_position)
```

```python
plt.figure(figsize=(8, 8))
plt.axis('equal')
plt.axis('off')
x, y = zip(*positions)
plt.plot(x, y, marker='.', markersize=5, linestyle='-', linewidth=2)
plt.title("Hilbert Curve of Order 3")
```
